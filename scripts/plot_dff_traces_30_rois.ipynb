# Plot dF/F traces
Loads dff.npy and plots 30 ROIs in separate rows.

```python
import numpy as np
import matplotlib.pyplot as plt

OUT_DIR = '/Users/megyoung/Desktop/mini2p Images/distdetrd/suite2p/plane0/custom_green_mask_traces'

dff = np.load(f'{OUT_DIR}/dff.npy')

rois_to_plot = np.arange(30)

fig, axes = plt.subplots(len(rois_to_plot), 1, figsize=(15, 1.5*len(rois_to_plot)), sharex=True)

if len(rois_to_plot) == 1:
    axes = [axes]

for ax, roi in zip(axes, rois_to_plot):
    ax.plot(dff[roi], lw=0.7)
    ax.set_ylabel(str(roi), rotation=0, labelpad=20)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

axes[-1].set_xlabel('Frame')
plt.suptitle('ΔF/F traces')
plt.tight_layout()
plt.show()
```
