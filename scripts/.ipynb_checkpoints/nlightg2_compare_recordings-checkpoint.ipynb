# nLightG2 — comparing recordings under identical baselines

Each recording becomes one whole-FOV ΔF/F per baseline definition. In multi-plane recordings, ΔF/F is computed per plane and then averaged across planes. Every recording uses the **same** definitions, set in seconds and percentiles, so they mean the same thing at 30 Hz (corridor2) and at about 7.5 Hz per plane (4-plane recordings).

Read the bottom-row panels first. Each line is one recording, plotted across baseline methods. If the lines keep their order, the difference between recordings (puff vs no puff, passive vs baseline) does not depend on the baseline choice. If they cross, it does.

Absolute ΔF/F across different fields of view, depths, zooms or laser powers is not directly comparable, so treat corridor2 vs passive as qualitative.

**Inputs are only read.** corridor2's Suite2p folders hold detection outputs (`F.npy` = ROI traces), so their whole-FOV traces are extracted into local `results/` folders first.

```python
import os
from pathlib import Path
import matplotlib.pyplot as plt
from IPython.display import display
from nlight import plotting, recording_comparison, suite2p_io, traces, workflows

plotting.use_lab_style()
MEG1 = Path(os.environ.get('NLIGHT_MEG1_ROOT', '/Volumes/mrsic_flogel/public/projects/SaEl_20220201_VIP/2pdata/LC/grab/meg1'))
LOCAL = Path(os.environ.get('NLIGHT_EXPORT_ROOT', '../results'))

# Suite2p roots (read only)
SOURCES = {
    'corridor2 passive viewing': MEG1 / 'corridor2/raw_data/suite2p',
    'corridor2 baseline': MEG1 / 'corridor2/baseline_data/suite2p',
}
# Where each recording's whole-FOV export lives. The 4-plane recordings already
# have exports in their Suite2p folders from the old notebook.
EXPORTS = {
    'corridor2 passive viewing': LOCAL / 'corridor2_raw_data',
    'corridor2 baseline': LOCAL / 'corridor2_baseline_data',
    'passive (4 planes)': MEG1 / 'passive/suite2p',
    'no puff (2x, 4 planes)': MEG1 / 'raw_data_2x/suite2p',
    'puff (2x, 4 planes)': MEG1 / 'raw_data_2x_puff/suite2p',
}
```

```python
# Extract corridor2 whole-FOV traces locally (skipped once done).
# Averaging the ~9 GB registered binary over the network takes several minutes.
SETTINGS = traces.TraceSettings(baseline_window_s=60, baseline_percentile=50)
for name, suite2p_root in SOURCES.items():
    for plane_dir in suite2p_io.plane_dirs(suite2p_root):
        if not (EXPORTS[name] / plane_dir.name / 'whole_image_roi_dff.npz').exists():
            print('extracting', name, plane_dir.name)
            traces.extract_whole_fov(plane_dir, SETTINGS, out_root=EXPORTS[name])
```

```python
SPEC = recording_comparison.BaselineSpec(current_window_s=60, current_percentile=50,
                                         fixed_percentiles=(10,), long_windows_s=(180, 300, 600))
SLOW_TIMESCALE_S = 10

def compare(names, title):
    summaries, table = workflows.compare_recordings({n: EXPORTS[n] for n in names}, SPEC, SLOW_TIMESCALE_S)
    fig = plotting.recording_comparison_figure(summaries, table, title=title)
    display(fig); plt.close(fig)
    cols = ['median_dff_pct', 'p5_to_p95_pp', 'slow_p5_to_p95_pp', 'fast_std_pp', 'raw_drift_pct']
    display(table[cols].round(2))
    if len(names) == 2:
        display(recording_comparison.paired_difference(table, *names).round(2))
    return summaries, table
```

## Airpuff: no puff vs puff (2x, 4 planes each)

```python
airpuff = compare(['no puff (2x, 4 planes)', 'puff (2x, 4 planes)'], 'Airpuff vs no airpuff')
```

## corridor2: passive viewing vs pre-corridor baseline

```python
corridor2 = compare(['corridor2 passive viewing', 'corridor2 baseline'], 'corridor2: passive viewing vs baseline')
```

## corridor2 (1 plane) vs passive (mean of 4 planes)

```python
across = compare(['corridor2 passive viewing', 'passive (4 planes)'], 'corridor2 vs passive')
```
